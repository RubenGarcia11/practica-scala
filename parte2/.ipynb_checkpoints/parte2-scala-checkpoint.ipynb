## Ejercicio 1 — Variables, tipos e inferencia

```scala
// Versión A: Declaración explícita
val nombre: String = "Juan"
var partidas: Int = 22
var puntuacion: Int = 60
var winRate: Double = 65.4
var activo: Boolean = true


println(s"$nombre     $partidas     $puntuacion     $winRate%     $activoA")
```

```scala
// Versión B: Scala infiere en los tipos
val nombre = "Alvaro"
var partidas = 28
var puntuacion = 45
var winRate = 22.3
var activo = false

println(s"$nombre     $partidas     $puntuacion     $winRate%     $activoA")
```

**Conclusión del Ejercicio 1:**
* **Tipos inferidos:** En la Versión B, Scala ha inferido `String` para `nombre`, `Int` para `partidas` y `puntuacion`, `Double` para `winRate` (por el decimal), y `Boolean` para `activo`.
* **Diferencias:** La versión B es más limpia y concisa de leer, ya que nos ahorramos escribir el tipo explícitamente, pero el compilador aplica el tipado estático igualmente de forma segura.
* **`val` vs `var`:** El nombre del jugador debería ser `val` (inmutable) porque normalmente un jugador no cambia de nombre a mitad de partida. Las partidas, puntuación, winrate y estado activo deberían ser `var` (mutables) porque irán cambiando conforme el jugador juegue más rondas.

## Ejercicio 2 — val, var y reasignación

```scala
val jugador = "Marta"
var puntuacion = 10

puntuacion += 5
puntuacion += 3

println(s"Puntuación final: $puntuacion")

jugador = "Herran"
```

**Conclusión del Ejercicio 2:**
* **Por qué `puntuacion` puede modificarse:** Porque fue declarada con `var` (variable), lo que permite reasignar su valor en memoria.
* **Por qué `jugador` no puede reasignarse:** Porque fue declarada con `val` (value), lo que la convierte en inmutable una vez asignada.
* **Mensaje de error:** Scala genera el error `error: reassignment to val`, indicando que el compilador prohíbe reescribir una constante.

```scala
val piDouble: Double = 3.14159265358979323846264338327
val piFloat: Float = 3.14159265358979323846264338327f

println(s"Valor Double: $piDouble")
println(s"Valor Float:  $piFloat")

val numEstudiantes: Int = 12
val haAprobado: Boolean = true
val mensaje: String = "Bien!"
```

## Ejercicio 4 — Función para determinar si una mano se pasa de 21

```scala
def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

println(bust(18))
println(bust(21))
println(bust(22))
println(bust(30))
```

## Ejercicio 5 — Comparación de dos manos


```scala
def maxHand(hand1: Int, hand2: Int): Int = {
  if (hand1 > hand2) {
    hand1
  } else {
    hand2
  }
}

println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

**Conclusión del Ejercicio 5:**
* **¿Qué sucede cuando ambas manos tienen el mismo valor?** La condición `hand1 > hand2` evalúa a `false`, por lo que el programa entra en el bloque `else` y devuelve `hand2`.
* **¿Es necesario modificarla?** No es estrictamente necesario modificarla. Como ambos valores son idénticos (ej. 21 y 21), devolver `hand2` es matemáticamente correcto porque el valor devuelto sigue siendo el máximo (21).

## Ejercicio 6 — Decidir el ganador de una partida

```scala
def ganador(handA: Int, handB: Int): Int = {
  if (bust(handA) && bust(handB)) {
    0
  } else if (bust(handA)) {
    handB
  } else if (bust(handB)) {
    handA
  } else {
    maxHand(handA, handB) 
  }
}

println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
```

**Documentación de las pruebas:**
* **26 y 20:** Cumple la condición `else if (bust(handA))` porque solo A (26) se pasa. Devuelve 20.
* **18 y 22:** Cumple la condición `else if (bust(handB))` porque solo B (22) se pasa. Devuelve 18.
* **24 y 25:** Cumple el primer `if (bust(handA) && bust(handB))` porque ambos superan 21. Devuelve 0.
* **17 y 19:** Ninguno se pasa. Llega al `else` final y usa `maxHand`. Devuelve 19.
* **21 y 20:** Ninguno se pasa. Llega al `else` final y usa `maxHand`. Devuelve 21.

## Ejercicio 7 — Arrays y mutabilidad

```scala
val jugadores = Array("Alex", "Chen", "Marta")

jugadores.foreach(println)

jugadores(0) = "Sindhu"

jugadores.foreach(println)

jugadores(1) = 500
```

**Conclusión del Ejercicio 7:**
* **Por qué cambian los elementos si es `val`:** `val` protege la referencia a la variable (no podemos hacer que `jugadoresArray` apunte a otro Array distinto), pero el objeto `Array` en sí mismo es una estructura de datos *mutable* por naturaleza en Scala, por lo que su contenido interno sí puede cambiar.
* **Por qué no se puede introducir 500:** Porque la inferencia de tipos detectó que es un `Array[String]`. Scala es fuertemente tipado, por lo que no permite insertar un `Int` en un array de cadenas de texto.
* **Diferencia entre reasignar y modificar:** Reasignar significa cambiar la caja entera por otra caja nueva (prohibido por `val`). Modificar un elemento significa abrir la caja y cambiar solo una pieza de su interior (permitido porque el Array es mutable).

## Ejercicio 8 — Creación e inicialización de Arrays

```scala
val puntuaciones = new Array[Int](4)

println(puntuaciones.mkString(", "))

puntuaciones(0) = 17
puntuaciones(1) = 24
puntuaciones(2) = 21
puntuaciones(3) = 19

println(puntuaciones.mkString(", "))
println(puntuaciones.length)
```

**Pregunta:** ¿Qué valores contiene un `Array[Int]` recién creado?
**Respuesta:** Contiene ceros (`0`) en todas sus posiciones. Scala inicializa los arrays numéricos con su valor por defecto (0 para Int/Double, `false` para Boolean, `null` para objetos).

## Ejercicio 9 — Recorrer un Array con while

```scala
val manosWhile = Array(17, 24, 21, 19, 26)
var i = 0

while (i < manosWhile.length) {
  val valorMano = manosWhile(i)
  println(s"$valorMano ${bust(valorMano)}")
  i = i + 1
}
```

**Conclusión del Ejercicio 9:**
* **Por qué el contador debe ser mutable:** Porque en cada iteración del bucle `while` necesitamos actualizar su valor (`i = i + 1`) para avanzar a la siguiente posición del array. Si fuera `val`, no podríamos incrementarlo.
* **Qué ocurriría si no incrementaras i:** Se produciría un **bucle infinito**. `i` siempre sería 0, que es menor que la longitud del array, por lo que el programa nunca saldría del bloque `while` e imprimiría el primer elemento eternamente.
* **Elementos de estilo imperativo:** El uso de una variable mutable (`var i`), la mutación explícita del estado (`i = i + 1`), y el hecho de dictar el "cómo" recorrer la estructura paso a paso mediante un bucle de control (`while`).

## Ejercicio 10 — Listas e inmutabilidad

```scala
val listaJugadores = List("Alex", "Chen", "Marta")
println(listaJugadores)

val nuevaLista = "Sindhu" :: listaJugadores
println(listaJugadores)
println(nuevaLista)

println(nuevaLista.length)
println(nuevaLista.reverse)
```

**Conclusión del Ejercicio 10:**
* **Por qué `::` produce una lista nueva:** La estructura `List` en Scala es estrictamente **inmutable**. Al ser inmutable, no dispone de métodos para modificar sus elementos en memoria. Cualquier operación de adición (`::`) o alteración devuelve siempre una lista completamente nueva, preservando la original intacta de forma segura.

## Ejercicio 11 — Construcción y concatenación de listas

## Ejercicio 12 — Operadores relacionales y lógicos

```scala
val handA = 18
val handB = 21
val handC = 25

val exp1 = handA > handB
val exp2 = handB == 21
val exp3 = handC != 21
val exp4 = handA <= 21 && handB <= 21
val exp5 = handA > 21 || handC > 21
val exp6 = !(handB > 21) // o también bust(handB) == false
```

| Expresión | Resultado | Explicación |
| :--- | :--- | :--- |
| `handA > handB` | false | 18 no es mayor que 21. |
| `handB == 21` | true | La variable `handB` vale exactamente 21. |
| `handC != 21` | true | 25 es distinto de 21. |
| `handA <= 21 && handB <= 21` | true | Ambos valores (18 y 21) son menores o iguales a 21. El operador AND requiere que ambas sean ciertas. |
| `handA > 21 \|\| handC > 21` | true | Aunque handA no supera 21, handC (25) sí. El operador OR requiere que al menos una sea cierta. |
| `!(handB > 21)` | true | `handB > 21` es falso, y el operador NOT (`!`) invierte el falso a verdadero. |

## Ejercicio 13 — foreach y funciones como valores

```scala
val manosEj13 = Array(17, 24, 21, 26, 18)

println("--- Versión A (while) ---")
var index = 0
while (index < manosEj13.length) {
println(s"Mano ${manosEj13(index)} se pasa: ${bust(manosEj13(index))}")
  index += 1
}
```

```scala
val manosEj13 = Array(17, 24, 21, 26, 18)

println("\n--- Versión B (foreach) ---")
manosEj13.foreach(mano => println(s"Mano $mano se pasa: ${bust(mano)}"))
```

**Conclusión del Ejercicio 13:**
* **¿Cuál necesita un contador?** La Versión A (`while`) necesita un contador explícito (`index`).
* **¿Cuál necesita una variable `var`?** La Versión A (`while`) requiere una variable mutable para alterar el índice en cada vuelta.
* **¿Cuál se aproxima más al estilo funcional?** La Versión B (`foreach`). Es más declarativa, nos centramos en el "qué" queremos hacer con cada elemento (pasar una función como valor) y delegamos el "cómo" recorrer la colección al propio lenguaje, sin usar estado mutable.

## Ejercicio 14 — Efectos secundarios y estilo de programación

```scala
var total = 0

def sumarAlTotal(valor: Int): Unit = {
  total = total + valor
}

sumarAlTotal(5)
println(total)
sumarAlTotal(10)
println(total)
```

```scala
def sumar(a: Int, b: Int): Int = {
  a + b
}
```

**Explicación:** La función `sumarAlTotal` modifica una variable situada fuera de su ámbito local usando una reasignación. Este es el **efecto secundario**: la función está mutando el estado global de la aplicación de forma "invisible" desde la perspectiva de sus parámetros de entrada y salida, lo que dificulta seguir el rastro de los datos.

| Característica | sumarAlTotal | sumar |
| :--- | :--- | :--- |
| **Modifica datos externos** | Sí | No |
| **Devuelve un resultado calculado** | No (devuelve Unit/nada) | Sí (devuelve Int) |
| **Utiliza efecto secundario** | Sí | No |
| **Estilo predominante** | Imperativo | Funcional |

## Ejercicio 15 — Programa integrado: torneo de Twenty-One

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)

def obtenerMejorJugador(nombres: List[String], puntuaciones: Array[Int]): String = {
  var mejorPuntuacion = -1
  var mejorJugador = ""
  var i = 0
  
  while (i < puntuaciones.length) {
    val puntos = puntuaciones(i)
    val nombre = nombres(i)
    
    if (!bust(puntos) && puntos > mejorPuntuacion) {
      mejorPuntuacion = puntos
      mejorJugador = nombre
    } else if (bust(puntos)) {
    } else {
    }
    i += 1
  }
  
  if (mejorPuntuacion != -1) s"$mejorJugador con $mejorPuntuacion puntos"
  else "Ningún jugador"
}

println("=== RESULTADOS RONDA 1 ===")
var i = 0
while (i < manos.length) {
  val nombre = jugadores(i)
  val puntos = manos(i)
  
  if (bust(puntos)) {
    println(s"- $nombre (Mano: $puntos) -> ¡Se ha pasado!")
  } else {
    println(s"- $nombre (Mano: $puntos) -> Puntuación válida.")
  }
  i += 1
}

println(s"\n>>> MEJOR PUNTUACIÓN RONDA 1: ${obtenerMejorJugador(jugadores, manos)}\n")

val manosRonda2 = Array(22, 19, 20, 21)

println("=== RESULTADOS RONDA 2 ===")
var j = 0
while (j < manosRonda2.length) {
  val nombre = jugadores(j)
  val puntos = manosRonda2(j)
  
  if (bust(puntos)) {
    println(s"- $nombre (Mano: $puntos) -> ¡Se ha pasado y no continúa!")
  } else if (puntos == 21) {
    println(s"- $nombre (Mano: $puntos) -> ¡Veintiuno exacto! Puntuación válida.")
  } else {
    println(s"- $nombre (Mano: $puntos) -> Puntuación válida.")
  }
  j += 1
}

println(s"\n>>> MEJOR PUNTUACIÓN RONDA 2: ${obtenerMejorJugador(jugadores, manosRonda2)}")
```

**Informe final del ejercicio 15:**

* **Qué partes de tu solución son mutables:** 
  Las variables de los contadores en los bucles `while` (`var i`, `var j`) y las variables que almacenan el estado temporal dentro de la función `obtenerMejorJugador` (`var mejorPuntuacion`, `var mejorJugador`).
* **Qué partes son inmutables:** 
  Las colecciones de datos originales (`val jugadores`, `val manos`, `val manosRonda2`) y los valores temporales dentro de los bucles que no necesitan cambiar una vez extraídos (`val nombre`, `val puntos`).
* **Dónde utilizas funciones:** 
  Utilizo la función `bust` (definida en el ej. 4) para verificar si una mano supera los 21 puntos, y una función propia llamada `obtenerMejorJugador` para encapsular la lógica de buscar el máximo valor válido en las colecciones.
* **Dónde utilizas estructuras de control:** 
  Utilizo bucles `while` para recorrer los índices de los Arrays y Listas simultáneamente, y bloques `if / else if / else` tanto dentro de la función propia como en el recorrido de la Ronda 2 para clasificar los resultados.
* **Qué parte de tu solución consideras más cercana al estilo funcional:** 
  El hecho de no modificar nunca la lista `jugadores` ni los arrays de puntuaciones originales al procesarlos. Además, la función `bust` es una función "pura" (solo depende de sus entradas y no tiene efectos secundarios).
* **Qué parte consideras más cercana al estilo imperativo:** 
  El uso de los bucles `while` con contadores manuales (`i += 1`) y la modificación constante de las variables internas `mejorPuntuacion` y `mejorJugador` para encontrar el ganador.
